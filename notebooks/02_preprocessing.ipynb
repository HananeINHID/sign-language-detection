# 02 — Preprocessing & Normalization

Loads extracted .npy landmark files, normalizes coordinates, verifies quality.

**Input:** `Drive/SignLanguage_Project/01_Landmarks/`  
**Output:** In-memory arrays, ready for export in notebook 03

```python
from google.colab import drive
drive.mount('/content/drive')
import sys
sys.path.append('/content/sign-language-pipeline')

from src.pipeline_utils import CLASSES, LABEL_MAP
from src.data_quality import run_full_quality_report
import numpy as np, os
```

```python
DRIVE_BASE = '/content/drive/MyDrive/SignLanguage_Project'
LAND_DIR   = f'{DRIVE_BASE}/01_Landmarks'

# Load all .npy files
X, y = [], []
for label in CLASSES:
    class_dir = os.path.join(LAND_DIR, label)
    if not os.path.isdir(class_dir): continue
    for fname in os.listdir(class_dir):
        if fname.endswith('.npy'):
            seq = np.load(os.path.join(class_dir, fname))
            X.append(seq)
            y.append(LABEL_MAP[label])

X = np.array(X)
y = np.array(y)
print(f'Loaded: X={X.shape}, y={y.shape}')
```

```python
# Normalize: clip extreme values, then min-max per feature
# (coordinates are already 0-1 from MediaPipe, but pose visibility can vary)
X = np.clip(X, 0, 1)
print('Normalization done.')
```
