```python
import json
import os
import shutil

# 1. Chemins sources (provenant de ton étape précédente)
dataset_path = "/kaggle/input/wlasl-processed"
json_path = os.path.join(dataset_path, 'WLASL_v0.3.json')
videos_src_dir = os.path.join(dataset_path, 'videos')

# 2. Chemin de destination sur ton DRIVE (ton repo cloné)
# Vérifie bien que ce chemin existe sur ton Drive
drive_output_dir = "/content/drive/MyDrive/Projets/sign-language-detection/data/raw"

# 3. Tes 5 mots cibles pour le premier sprint
mots_cibles = ['eat', 'drink', 'water', 'sleep', 'medicine']

print("⏳ Début du tri et de la copie vers le Drive...")

# Chargement du JSON
with open(json_path, 'r') as f:
    wlasl_data = json.load(f)

total_copies = 0

for entry in wlasl_data:
    gloss = entry['gloss'] # Le mot en anglais

    if gloss in mots_cibles:
        # Créer le dossier du mot dans le Drive (ex: data/raw/eat)
        target_folder = os.path.join(drive_output_dir, gloss)
        os.makedirs(target_folder, exist_ok=True)

        for instance in entry['instances']:
            video_id = instance['video_id']
            file_name = f"{video_id}.mp4"
            src_file = os.path.join(videos_src_dir, file_name)
            dst_file = os.path.join(target_folder, file_name)

            # Copie physique vers le Drive
            if os.path.exists(src_file):
                shutil.copy(src_file, dst_file)
                total_copies += 1

print(f"✅ Terminé ! {total_copies} vidéos ont été copiées dans ton Drive.")
print(f"📂 Vérifie ici : {drive_output_dir}")
```
